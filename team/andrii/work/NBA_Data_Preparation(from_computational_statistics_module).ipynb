## MA2502 - Preparation of the NBA data 

In this notebook, we perform various operations on the raw data set `NBA_Shots_Raw.csv` in order to prepare it for our analyses. The cleaning and preprocessing of the data is a very important step when dealing with real-life data. We also illustrate how to enrich the data using the auxiliary data sets `Player_Info.csv` and `Player_Salary.csv`. 

The raw data set `NBA_Shots_Raw` contains $N_{\mathrm{raw}}=128{,}069$ observations of $23$ variables (or features); these variables are listed below. 

Variable | Description
:--- | :--- | 
`GAME_ID` | Unique id number of the game.
`DATE` | Date the game took place.
`HOME_TEAM` | Team playing at home. 
`AWAY_TEAM` | Team playing away from home. 
`PLAYER_NAME` | Name of the shooting player. 
`PLAYER_ID` | Unique ID number of the shooting player.
`LOCATION` | Whether the player was on the home (H) or away (A) team.
`W` | Whether the player's team won (W) or lost (L) the game.
`FINAL_MARGIN` | The final margin for the player's team (negative means defeat).
`SHOT_NUMBER` | The number of the shot taken by the shooting player in that game.
`PERIOD` | The period of the game that the shot was taken. A value of 5 or greater indicates that the game was tied after 4 quarters, and that the shot was taken in an overtime period.
`GAME_CLOCK` | Time display on the game clock when the shot was taken. 
`SHOT_CLOCK` | The time remaining on the shot clock (the time limit in the NBA is 24 seconds). 
`DRIBBLES` | Number of dribbles by the player before the shot was taken. 
`TOUCH_TIME` | The time that the ball was in the shooting player's hand. 
`SHOT_DIST` | The distance of the shooting player from the basket. 
`PTS_TYPE` | 2 for shots from inside the arc, 3 for shots from outside the arc. 
`SHOT_RESULT` | "made" or "missed". 
`CLOSEST_DEFENDER` | Name of the nearest defender when the shot was taken. 
`CLOSEST_DEFENDER_ID` | Unique id number of the nearest defender. 
`CLOSE_DEF_DIST` | Distance of the nearest defender when the shot was taken. 
`FGM` | Equal to 1 if the shot was made (scored) otherwise 0. 
`PTS` | The number of points scored with the shot. 

The free throws were not recorded (i.e. there are no free throws in the data set).

The variable `GAME_CLOCK` is recorded as a string. To turn it into a more exploitable variable, from `GAME_CLOCK`, we create an new variable `GAME_CLOCK_SEC` that indicates the number of seconds remaining before the end of the period. 

```
# Convert GAME_CLOCK into seconds (i.e. remaining second before the end of the period)

# Example
#as.matrix(read.table(text = NBA$GAME_CLOCK[1:10], sep = ":")) %*% c(60,1)

GAME_CLOCK_SEC <- as.matrix(read.table(text = NBA$GAME_CLOCK, sep = ":")) %*% c(60,1)


options(repr.plot.width=7, repr.plot.height=5)

hist(GAME_CLOCK_SEC, freq=FALSE, col='purple')
```

```
NBA$GAME_CLOCK[1:2]
as.matrix(read.table(text = NBA$GAME_CLOCK[1:2], sep = ":")) %*% c(60,1)
```

The shot clock does not start when the time remaining in the period is less than 24 second; this means that we should be able to correct many missing values in `SHOT_CLOCK` by replacing them by the remaining time in the period. Obviously, we need to carefully check when this can or cannot be done. 

We start by identifying the missing values in `SHOT_CLOCK` that corresponds to entries in `GAME_CLOCK_SEC` that are less than 24 seconds. 

```
print(summary(GAME_CLOCK_SEC[SC_NA]))

print(sum(GAME_CLOCK_SEC[SC_NA]<= 24))
```

We can then identify the missing values in `SHOT_CLOCK` that we can actually obtain from `GAME_CLOCK_SEC`. 

```
Ind_SC_NA_correct_IN <- (1:nb_SC_NA)[GAME_CLOCK_SEC[SC_NA] <= 24]
Ind_SC_NA_correct <- SC_NA[Ind_SC_NA_correct_IN]

print(length(Ind_SC_NA_correct))
```

```
Ind_Missing_SC <- SC_NA[-Ind_SC_NA_correct_IN]
print(length(Ind_Missing_SC))

print(summary(GAME_CLOCK_SEC[Ind_Missing_SC]))
```

The following rows correspond to entries of the data set for which we do not have values for `SHOT_CLOCK` (even after our correction).

Correction of the entries of `SHOT_CLOCK` that can be replaced by `GAME_CLOCK_SEC`.

## 4. Summary of the performed operations 

### 4.1. Cleaning of the rows

1. Correction of the misspelled names in `PLAYER_NAME`. 

2. Removal of the outliers in `TOUCH_TIME`. 

3. Removal/correction of the outliers in `PTS_TYPE`, i.e. 2-pointers with `SHOT_DIST` larger than 23.75, and 3-pointers with `SHOT_DIST` smaller than 22 (see the figure below). 

4. Correction of the variable `SHOT_CLOCK` (value replaced by the period-remaining-time when below 24 sec.) and removal of the entries that could not be corrected.  

**Remark:** The variable `CLOSE_DEF_DIST` contains a few large values (the maximum is 53.2 while the median is 3.7). The length of a NBA court is 94 ft, so that although unusual,  such situation is realistic: it may for instance occur during "extreme counter attacks", or when the defending team stopped playing. The variable was kept untouched.  


### 4.2. About the columns 

1. Removal of the columns `SHOT_RESULT` and `PTS`. 

2. Replacement of the variable `GAME_CLOCK` (string of the type "10:03") by the variable `SEC_REMAIN` (integer) indicating the ***number of seconds before the end of the period*** (this variable was used to correct `SHOT_CLOCK` when possible). 

3. Renaming of the column `W` by `WIN_LOSE`. This variable is a 2-level categorical variables returning "W" if the shooter is part of the team which won the game, and "L" is his team lost. 

4. Renaming of the column `FGM` by `SUCCESS`. This variable is a 2-level categorical variables returning "1" is the shot was made, and "0" otherwise. 

The number of column after the data preparation is $21$ (against $23$ in the original data set). 

The file containing the new data set is named `NBA_Shots_Clean.csv`.
